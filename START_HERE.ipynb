## 🧭 Wie funktioniert so ein Notebook?

Ein Notebook ist eine Liste von **Zellen**. Es gibt zwei Sorten:

- **Text-Zellen** (wie diese) erklären etwas.
- **Code-Zellen** enthalten Python-Code. Sie erkennen sie an `In [ ]:` links.

**Das Wichtigste – so führst du eine Zelle aus:**

1. Klicke in die Zelle.
2. Drücke **Shift + Enter** (oder den ▶-Play-Button oben).

Probiere es gleich bei der Setup-Zelle unten aus! Unter der Zelle erscheint dann eine Nummer (`In [1]:`) und die Ausgabe.

**Drei Rettungsanker, wenn etwas seltsam ist:**
- **Kernel → Restart Kernel**: Python "vergisst" alles und startet frisch (bei komischen Fehlern oft die Lösung).
- **Run → Run All Cells**: führt das ganze Notebook von oben nach unten aus.
- **File → Save** und danach einfach weitermachen. Nichts kann kaputtgehen – du kannst jedes Notebook neu vom Repository holen.

**Hilfe bei Fehlern:** `docs/troubleshooting.md` – und: Fehler sind **Teil des Lernens**. Wir erklären sie dir, sobald sie auftauchen.
