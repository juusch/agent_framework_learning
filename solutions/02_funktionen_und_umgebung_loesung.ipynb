## 🧠 Konzept: Funktion – das Rezept mit Namen

**WARUM:** Ohne Funktionen müsstest du dieselben Zeilen endlos kopieren. Später in dieser Akademie wirst du **dieselbe** Funktion rufen: einmal schreibst du `beantworte(frage)` – hundertmal benutzt du sie.

**WAS:** Eine Funktion ist ein benannter, wiederverwendbarer Anweisungsblock. Wie ein **Kaffeevollautomat**: Knopf drücken (Aufruf), Zutaten wählen (Parameter), Getränk erhalten (**Rückgabewert**).

**WIE / TRY:**

````python
def begruessung(name, stil="hoeflich"):
    """Gibt eine Begruessung zurueck (Docstring = Bedienungsanleitung)."""
    if stil == "locker":
        return f"Na, {name}! Was geht?"
    return f"Guten Tag, {name}."

nachricht = begruessung("Anna")
print(nachricht)
print(begruessung("Ben", stil="locker"))
````

### 🔎 Was sehen wir hier?
- `stil="hoeflich"` ist ein **Standardwert** – die Funktion funktioniert auch ohne zweites Argument.
- `return` **gibt** einen Wert zurück (das Getränk). `print` würde ihn nur *anzeigen* – die Funktion wäre dann leer.

**Der wichtigste Unterschied der Lektion:** `return` = zurückgeben (damit andere Code weiterrechnen kann), `print` = nur auf den Bildschirm. Merksatz: **return rechnet, print plaudert.**

````python
def durchschnitt_falsch(zahlen):
    print(sum(zahlen) / len(zahlen))       # nur Anzeige -> Rueckgabewert ist None!

def durchschnitt_richtig(zahlen):
    if not zahlen:                         # leer -> 0 statt Absturz!
        return 0
    return sum(zahlen) / len(zahlen)

ergebnis = durchschnitt_falsch([2, 4])     # gibt 3.0 AUS, aber...
print("ergebnis ist:", ergebnis)          # ...die Variable enthaelt None
print(durchschnitt_richtig([2, 4, 9]))
print(durchschnitt_richtig([]))            # keine DivisionDurchNull-Panik
````

### 🛑 Fehler sind Freunde (3): ZeroDivisionError & None

`durchschnitt_falsch([])` würde mit `ZeroDivisionError` abstürzen (Division durch `len([]) = 0`). Die **Absicherung** (`if not zahlen: return 0`) nennt man *defensive Programmierung*: Unerwartete Eingaben abfangen, bevor sie Schaden anrichten. Genau das wird dein Assistent später bei Nutzerfragen tun.

## ✏️ Deine Aufgabe: Robuste Durchschnitts-Funktion

Du schreibst eine Funktion, die **niemals abstürzt** – egal, was hineinkommt.

**Anforderungen:**

- Schreibe `durchschnitt(zahlen)`, die den Durchschnitt einer Zahlenliste zurückgibt (`return`!).

- Bei einer **leeren** Liste gibt sie `0` zurück (kein Absturz).

- Schreibe zusätzlich `ist_arbeitszeit(stunde)`: gibt `True` zurück, wenn die Stunde zwischen 9 und 17 liegt.

**Schreibe deinen Code in die folgende Zelle:**

````python
def durchschnitt(zahlen):
    if not zahlen:
        return 0
    return sum(zahlen) / len(zahlen)

def ist_arbeitszeit(stunde):
    return 9 <= stunde <= 17
````

**💡 Warum diese Loesung funktioniert:**

Erst den Sonderfall (leere Liste) absichern, dann den Normalfall berechnen – dieses Muster ("Erst prüfen, dann rechnen") zieht sich durch die ganze Akademie. `9 <= stunde <= 17` ist eine Python-Kette: sie prüft beide Bedingungen auf einmal.

- [ ] `ist_arbeitszeit(12)` ist `True`, `ist_arbeitszeit(3)` ist `False`.

- [ ] Beide Funktionen nutzen `return` (nicht nur `print`).

**🧪 Fuehre die folgende Zelle aus, um dein Ergebnis zu pruefen:**

````python
assert durchschnitt([2, 4, 9]) == 5.0, "Durchschnitt von [2,4,9] sollte 5.0 sein."
assert durchschnitt([]) == 0, "Leere Liste soll 0 liefern, nicht abstuerzen!"
assert ist_arbeitszeit(12) is True
assert ist_arbeitszeit(3) is False
print("Alle Prüfungen bestanden! ✅")
````

## ✏️ Deine Aufgabe: Zusammenfassungs-Werkstatt

Du schreibst eine Funktion, die eine Liste von Zahlen analysiert – und auch bei Unsinn nicht abstürzt.

**Anforderungen:**

- Schreibe `analyse(zahlen)`, das ein Dictionary zurückgibt mit den Schlüsseln `"anzahl"`, `"durchschnitt"` und `"maximum"`.

- Bei einer leeren Liste: `{"anzahl": 0, "durchschnitt": 0, "maximum": 0}` – ohne Absturz.

- Nutze in der Funktion `try/except` mindestens einmal (z. B. für den Leerfall oder für die Division).

**Schreibe deinen Code in die folgende Zelle:**

````python
def analyse(zahlen):
    try:
        return {
            "anzahl": len(zahlen),
            "durchschnitt": sum(zahlen) / len(zahlen),
            "maximum": max(zahlen),
        }
    except (ZeroDivisionError, ValueError):   # leere Liste: beide moeglich
        return {"anzahl": 0, "durchschnitt": 0, "maximum": 0}
````

**💡 Warum diese Loesung funktioniert:**

Der `try`-Block enthält den Normalfall; wenn die Liste leer ist, löst `sum/len` bzw. `max` einen Fehler aus, den `except` in ein sauberes Ergebnis-Dict übersetzt. Die Funktion kann damit **nie** abstürzen – genau das wirst du von deinem Assistenten später auch verlangen.

### ✅ Richtig, wenn:

- [ ] `analyse([2, 4, 9])` liefert `anzahl=3`, `durchschnitt=5.0`, `maximum=9`.

- [ ] `analyse([])` liefert alle Werte als 0 – kein Crash.

**🧪 Fuehre die folgende Zelle aus, um dein Ergebnis zu pruefen:**

````python
ergebnis = analyse([2, 4, 9])
assert ergebnis["anzahl"] == 3 and ergebnis["maximum"] == 9
assert abs(ergebnis["durchschnitt"] - 5.0) < 0.001
leer = analyse([])
assert leer == {"anzahl": 0, "durchschnitt": 0, "maximum": 0}
print("Alle Prüfungen bestanden! ✅")
````
