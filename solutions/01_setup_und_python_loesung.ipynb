## 🧠 Konzept: Die Variable – ein beschrifteter Kasten

**WARUM:** Programme müssen sich Dinge merken: deinen Namen, eine Frage, eine Antwort. Ohne Gedächtnis wäre jedes Programm eine Taschenrechner-Eingabe, die sofort vergisst.

**WAS:** Eine Variable ist ein **Name für einen Wert**. Denk an einen beschrifteten Kasten: Du legst etwas hinein (`name = "Anna"`) und kannst später jederzeit hineinschauen (`name`).

**WIE:** Mit `=` – Vorsicht: Das ist keine Mathematik-Gleichung, sondern eine **Zuweisung**: "Lege das rechts Stehende in den Kasten links."

```python
name = "Anna"
alter = 30
groesse = 1.68

print(name, "ist", alter, "Jahre alt.")
```

### 🔎 Was sehen wir hier?
`"Anna"` ist ein **String** (Text), `30` ein **Integer** (ganze Zahl), `1.68` ein **Float** (Dezimalzahl). Das nennt man **Typen**. Python erkennt sie automatisch – du musst sie nur nicht mischen, ohne es zu merken.

### 🛑 Fehler sind Freunde (1): TypeError

Führe die nächste Zelle aus – sie **wird absichtlich fehlschlagen**. Das ist gewollt!

```python
# Diese Zelle provoziert absichtlich einen Fehler – lies die Meldung genau:
try:
    print("Alter naechstes Jahr: " + alter)
except TypeError as fehler:
    print("TypeError:", fehler)
    print("-> Python weigert sich, Text und Zahl mit + zu verkleben.")
```

**Was ist passiert?**
- `TypeError: can only concatenate str (not "int") to str`
- **Bedeutung:** Du hast versucht, Text und Zahl mit `+` zu verkleben. Python weigert sich – zu Recht.
- **Lösung:** f-Strings! Sie betten Variablen jeden Typs sauber in Text ein:

```python
print(f"Alter naechstes Jahr: {alter + 1}")
print(f"{name} ist {groesse} m gross.")
```

Das `f` vor den Anführungszeichen macht den String zum **f-String**: Alles in `{...}` wird durch den Wert ersetzt – und dort drin darf sogar gerechnet werden (`{alter + 1}`).

## 🧠 Konzept: Liste – die nummerierte Schublade

**WARUM:** Einzelne Variablen werden unpraktisch, wenn du **viele** Dinge desselben Typs hast: 3 Artikel, 100 Fragen, 500 Dokumente.

**WAS:** Eine Liste ist eine geordnete Sammlung. Wichtig: Das erste Element hat den Index **0** – zähle wie ein Computer!

**WIE / TRY:**

```python
einkauf = ["Milch", "Brot", "Aepfel"]
print("Komplett:", einkauf)
print("Erstes Element (Index 0):", einkauf[0])
print("Anzahl Elemente:", len(einkauf))

einkauf.append("Butter")   # hinten anhaengen
print("Nach append:", einkauf)
```

### Schleifen: mit jedem Element etwas tun

**WARUM:** Bei 500 Dokumenten willst du nicht 500 Zeilen schreiben. Die Schleife sagt: "Für **jedes** Element: tue Folgendes."

```python
gesamt = 0
for artikel in einkauf:
    print("- Artikel:", artikel)

for zahl in [2, 4, 9]:
    gesamt = gesamt + zahl
print("Summe:", gesamt)
```

## 🧠 Konzept: Dictionary – das Wörterbuch

**WARUM:** Listen kennen nur Positionen ("Element 0"). Oft willst du mit **Namen** nachschlagen: Was kostet Milch? Was ist Annas Rolle?

**WAS:** Ein Dictionary (`dict`) speichert **Schlüssel → Wert**-Paare wie ein Wörterbuch: Stichwort suchen, Erklärung finden.

**WIE / TRY:**

```python
preise = {"Milch": 1.20, "Brot": 2.50, "Aepfel": 3.00}

print("Preis von Milch:", preise["Milch"])
print("Sicher nachschlagen (mit Standardwert):", preise.get("Kaviar", "unbekannt"))

preise["Kaffee"] = 4.99    # neues Paar anlegen
print("Alle Paare:", preise)
```

### 🛑 Fehler sind Freunde (2): KeyError

`preise["Kaviar"]` würde mit `KeyError` abstürzen, weil der Schlüssel fehlt. `.get(schluessel, standardwert)` ist die sichere Variante: Sie liefert nie einen Absturz, sondern den Standardwert. Merke: **Klammern `[...]` = streng, `.get(...)` = gnädig.**

## 🧠 Konzept: Bedingungen – das Programm entscheidet

**WARUM:** Dein späterer Knowledge Assistant wird entscheiden: "Das ist eine Verbindungsfrage → nutze den Graphen!" Das Grundmuster lernst du jetzt.

**WIE / TRY:**

```python
alter = 30

if alter >= 18:
    print("Volljaehrig")
elif alter > 12:
    print("Jugendlich")
else:
    print("Kind")

# Vergleiche: ==  gleich   !=  ungleich   >  <  >=  <=
print("Ist 30 gleich 30?", 30 == 30)
```

Das Ergebnis eines Vergleichs ist ein **Boolean**: `True` oder `False` – der ja/nein-Typ, den du später überall brauchst.

## ✏️ Deine Aufgabe: Dein Einkaufslisten-Rechner

Du kombinierst Liste, Dictionary und Schleife zu einem echten Mini-Programm.

**Anforderungen:**

- Erstelle eine Liste `mein_einkauf` mit mindestens 3 Artikeln.

- Erstelle ein Dictionary `meine_preise`, das zu jedem Artikel einen Preis (Zahl) speichert.

- Berechne in einer Schleife die Summe aller Preise der Liste und speichere sie in `summe`.

- Gib die Summe mit zwei Nachkommastellen aus (Tipp: `f"{summe:.2f}"`).

**Schreibe deinen Code in die folgende Zelle:**

```python
mein_einkauf = ["Milch", "Brot", "Aepfel"]
meine_preise = {"Milch": 1.20, "Brot": 2.50, "Aepfel": 3.00}

summe = 0
for artikel in mein_einkauf:
    summe += meine_preise[artikel]

print(f"Summe: {summe:.2f} Euro")
```

**💡 Warum diese Loesung funktioniert:**

Die Schleife läuft über die Liste, schlägt pro Artikel den Preis im Dictionary nach und erhöht `summe`. Genau dieses Muster ("über eine Liste iterieren, Werte nachschlagen, akkumulieren") wirst du in Build 1 beim Dokumenten-Scoring wiedersehen!

### ✅ Richtig, wenn:

- [ ] `summe` ist eine Zahl und stimmt (rechne per Hand nach!).

- [ ] Die Ausgabe zeigt die Summe mit 2 Nachkommastellen.

- [ ] Du hast eine `for`-Schleife verwendet.

**🧪 Fuehre die folgende Zelle aus, um dein Ergebnis zu pruefen:**

```python
assert len(mein_einkauf) >= 3, "Die Liste braucht mindestens 3 Artikel."
assert len(meine_preise) >= 3, "Das Dictionary braucht mindestens 3 Preise."
erwartung = sum(meine_preise[a] for a in mein_einkauf if a in meine_preise)
assert abs(summe - erwartung) < 0.001, f"summe sollte {erwartung} sein, ist aber {summe}."
print("Alle Prüfungen bestanden! ✅")
```

</details>
